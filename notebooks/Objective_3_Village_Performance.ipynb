## Functions

```python
def summary(df):
  'function to generate summary statistics for each dataset'
  display(df.head())
  print('\n\nInfo: \n')
  display(df.info())
  print('\nNumeric Value Summary: \n')
  display(df.describe())
  str_cols = df.select_dtypes(include = ['object','bool'])
  print('\nString Column Summary: \n')
  for c in str_cols:
    print(f'\n\n**{c}**:\n')
    print('Number Unique Values: ')
    display(df[c].nunique())
    print('\nMost Common Values: ')
    display(df[c].value_counts().nlargest(10))
```

## Zones By Revenue (Done)

```python
# check to see if total paid per contract in repayment is the same as total paid per contract for reconciled payment
reconcile_contracts = reconciled_payment.groupby('Destination Contract Reference',as_index= False)['Amount'].sum()
repayment_contracts = repayment_data.groupby('Contract Reference', as_index = False)['Amount Paid'].sum()
data_check = reconcile_contracts.merge(repayment_contracts, left_on = 'Destination Contract Reference', right_on = 'Contract Reference', how = 'right')
payment_difference = abs(data_check['Amount']*(-1) - data_check['Amount Paid'])
payment_difference.sum()
#it is not so we will use reconcile contracts to measure revenue, because it contains more records
```

```python
#removing because it's mostly empty
reconciled_payment.drop(columns = ['Client Id'], inplace = True)
```

```python
#Create a dataframe matching up reconcile_payments with villages
rec_w_client = reconciled_payment.merge(repayment_data[['Contract Reference', 'Client Id']], left_on = 'Destination Contract Reference', right_on = 'Contract Reference', suffixes = ("",""))
rec_w_village = rec_w_client.merge(client_data[['Id', 'L0 Entity Name']], left_on = 'Client Id', right_on = 'Id', suffixes= ("", " Client"))
rec_w_village.rename(columns = {'L0 Entity Name': 'Village', 'Amount': 'Revenue'}, inplace = True)
rec_w_village['Date'] = pd.to_datetime(rec_w_village['Date'])
rec_w_village.head()
```

```python
#changing amount paid to positive value
rec_w_village['Revenue'] = rec_w_village['Revenue']*(-1)
```

```python
rec_w_village.to_csv('reconcile_payments_village.csv')
```

```python
#Find total revenue Grouped by Village
village_revenue = rec_w_village.groupby('Village', as_index = False)['Revenue'].sum()
village_revenue.sort_values('Revenue', ascending = False).head()
```

```python
#Rank Villages based on total revenue
village_revenue['Rank'] = village_revenue['Revenue'].rank(method = 'min', ascending = False)
village_revenue.sort_values('Rank')
```

## Zone By Revenue Per Contract (Done)

```python
#Rank Villages based on total revenue per contract
num_village_contracts = rec_w_village.groupby('Village', as_index = False)['Contract Reference'].nunique().rename(columns = {'Contract Reference': 'Number Contracts'})
avg_village_revenue = village_revenue.merge(num_village_contracts, on = 'Village')
avg_village_revenue['Average Revenue'] = avg_village_revenue['Revenue']/avg_village_revenue['Number Contracts']
avg_village_revenue['Rank'] = avg_village_revenue['Average Revenue'].rank(method = 'dense', ascending = False)
display(avg_village_revenue.sort_values('Rank'))
```

```python
total_retention = (len(client_data) - len(client_data[client_data['Has Defaulted Contracts'] == True]))/len(client_data)
total_retention
```

```python
#retention by village
village_clients = client_data.groupby('L0 Entity Name')['Id'].count()
village_defaults = client_data[client_data['Has Defaulted Contracts'] == True].groupby('L0 Entity Name')['Id'].count()
village_retention = (village_clients - village_defaults)/village_clients
village_retention = village_retention.reset_index().rename(columns = {'Id':'Retention Rate', 'L0 Entity Name': 'Village'}).fillna(0)
village_retention['Rank'] = village_retention['Retention Rate'].rank(method = 'dense', ascending = False)
village_retention.sort_values('Rank')
```

```python
#find the proportion of clients who have not had their panel repossesed by village
repossessed_client = stock_movement[(~stock_movement['Origin Client Id'].isna())&((stock_movement['Note'] == '[Automatic] Repossession')|(stock_movement['Note'] == '[Automatic] Repossesion'))]['Origin Client Id'].to_list()
type(repossessed_client)
```

```python
client_data['Repossessed'] = client_data['Id'].isin(repossessed_client)
```

```python
village_repo = client_data[client_data['Repossessed'] == False].groupby('L0 Entity Name')['Id'].count()
village_retention2 = (village_clients - village_repo)/village_clients
village_retention2 = village_retention2.reset_index().rename(columns = {'Id':'Repossession Rate', 'L0 Entity Name': 'Village'}).fillna(0)
village_retention2['Rank'] = village_retention2['Repossession Rate'].rank(method = 'min', ascending = True)
village_retention2.sort_values('Rank').head(10)
```

## Zones by Lead Conversion (Done)

## Zones with best payment consistency (Done)

```python
#first merge client, and bill datasets
area_bill = client_data[['Id','L0 Entity Id', 'L0 Entity Name']].merge(repayment_data, left_on = 'Id', right_on = 'Client Id', how = 'right')
area_bill.head()
#display(area_bill_data.columns)

#changing column name for clarity
area_bill.rename(columns = {'L0 Entity Name': 'Village'}, inplace= True)

#changing payment date and expected payment date to datetime
area_bill['Expected Payment Date'] = pd.to_datetime(area_bill['Expected Payment Date'])
area_bill['Payment Date'] = pd.to_datetime(area_bill['Payment Date'])

#looking at na values in 'Days Late' column
na_days_late = area_bill[area_bill['Days Late'].isna()]
print(len(na_days_late))
display(na_days_late.head())
display(na_days_late['Type'].value_counts())
# NA's are all downpayments or reimbursment for downpayments so these can be safely ignored
```

```python
downpayments = area_bill[area_bill['Type'] == 'Downpayment']
downpayments.head()
```

```python
downpayments.to_csv('downpayments.csv')
```

```python
#removing ['Downpayment', 'Reversed Contract Payment', 'Offer Change Delay', 'Manual Discount', 'Reversed Downpayment'] Type because they are not payments
area_bill = area_bill[~area_bill['Type'].isin(['Downpayment', 'Reversed Contract Payment', 'Offer Change Delay', 'Manual Discount', 'Reversed Downpayment'])]
```

```python
# where are people on time most often?
on_time_payments = area_bill[area_bill['Days Late'] <= 0]
len(on_time_payments)
on_time = on_time_payments.groupby('Village', as_index = False)['Days Late'].count().rename(columns = {'Days Late':'Number On Time'})
num_village_payments = area_bill.groupby('Village', as_index = False)['Days Late'].count().rename(columns = {'Days Late':'Number Payments'})
on_time = on_time.merge(num_village_payments, on = 'Village')
on_time['Proportion'] = on_time['Number On Time']/on_time['Number Payments']
on_time['Rank'] = on_time['Proportion'].rank(method = 'dense', ascending = False)
display(on_time.sort_values('Rank'))

display(on_time[on_time['Village'] == 'Wirmana'])

print('Median Days Early by Village: \n', on_time_payments.groupby('Village')['Days Late'].median().nsmallest(4)*(-1))
```

```python
on_time_payments.to_csv('on_time_granular.csv')
on_time.to_csv('on_time_agg.csv')
```

```python
# where were people late most often?
late_payments = area_bill[area_bill['Days Late'] > 0]
late = late_payments.groupby('Village', as_index = False)['Days Late'].count().rename(columns = {'Days Late':'Number Late'})
late = late.merge(num_village_payments, on = 'Village')
late['Proportion'] = late['Number Late']/late['Number Payments']
late['Rank'] = late['Proportion'].rank(ascending = False)
len(late_payments)
display(late.sort_values('Rank').head(10))

print('Median Days Late by Village: \n', late_payments.groupby('Village')['Days Late'].median())
```

```python
late_payments.to_csv('late_granular.csv')
late.to_csv('late_agg.csv')
```
